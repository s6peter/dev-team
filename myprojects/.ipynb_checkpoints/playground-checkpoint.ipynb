```python
import string

def password_strength_counter(password):
    strength = {
        'length': len(password) >= 8,
        'digit': any(char in string.digits for char in password),
        'lowercase': any(char in string.ascii_lowercase for char in password),
        'uppercase': any(char in string.ascii_uppercase for char in password),
    }
    
    if all(strength.values()):
        return f"{password} is strong"
    else:
        return f"{password} is weak. Try again"

# Example usage
password = "57yyG58oo"
print(password_strength_counter(password))
```

```python
import string

def password_strength_counter(password):
    strength_checks = {
        'length': len(password) >= 8,
        'digit': any(char in string.digits for char in password),
        'lowercase': any(char in string.ascii_lowercase for char in password),
        'uppercase': any(char in string.ascii_uppercase for char in password),
        'special': any(char in string.punctuation for char in password),
    }

    
    print(strength_checks )
    score = sum(strength_checks.values())
    
    if score == 5:
        return f"✅ {password} is strong! (Score: {score}/5)"
    else:
        suggestions = []
        if not strength_checks['length']:
            suggestions.append("make it at least 8 characters")
        if not strength_checks['digit']:
            suggestions.append("add a digit (0–9)")
        if not strength_checks['lowercase']:
            suggestions.append("include a lowercase letter")
        if not strength_checks['uppercase']:
            suggestions.append("include an uppercase letter")
        if not strength_checks['special']:
            suggestions.append("add a special character (!@#$...)")
        suggestion_text = "; ".join(suggestions)
        return f"⚠️ {password} is weak. (Score: {score}/5)\nSuggestions: {suggestion_text}"

# List of passwords
passwords = ["password", "Pa$$w0rd", "SuperSecurePwd!", "weakpw"]

# Check each password
for pwd in passwords:
    print(password_strength_counter(pwd))
    print("-" * 50)
```

```python
password = "MyP@ssword123"

has_digit = any(c.isdigit() for c in password)
has_upper = any(c.isupper() for c in password)
has_special = any(c in "!@#$%^&*" for c in password)
has_lower = any(c.islower() for c in password)

print(all([has_digit, has_upper, has_special,has_lower]))
# True if the password passes ALL checks

```

```python

def solution(listA):
    count_dict = {}
    for element in listA:
        count_dict[element] += 1
        if count_dict[element] > len(listA) // 2:
       # if count_dict[element] += 1 with defaultdict
            return element
    return -1

listA = [1,2,4,5,5,3,1]
solution(listA)
```
