```python
import os

def combine_files_in_directory(output_filename="combined_output.txt"):
    """
    Combines the content of all files in the current directory into a single output file.

    Args:
        output_filename (str): The name of the file to save the combined content to.
    """
    current_directory = os.getcwd()
    list_dir_items = os.listdir(current_directory)
    
    # List to store paths of regular files
    files_to_combine = []
    for item in list_dir_items:
        item_path = os.path.join(current_directory, item)
        if os.path.isfile(item_path) and item != output_filename: # Exclude the output file itself
            files_to_combine.append(item_path)
            
    if not files_to_combine:
        print("No files found to combine in the current directory.")
        return

    print(f"Combining {len(files_to_combine)} files into '{output_filename}'...")

    try:
        with open(output_filename, 'w') as outfile:
            for filepath in files_to_combine:
                with open(filepath, 'r') as infile:
                    outfile.write(infile.read())
                    outfile.write("\n\n--- End of file: " + os.path.basename(filepath) + " ---\n\n") # Optional separator
        print(f"\nSuccessfully combined all files into '{output_filename}'.")
        print(f"The combined file is located at: {os.path.join(current_directory, output_filename)}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Call the function to combine files
combine_files_in_directory()

with open("/home/peterg/python-Project/combined_output.txt", "r") as f:
    content = f.readlines(1000)
    print(content)
```

```python
 import os

# Get the current working directory
print(f"Current working directory: {os.getcwd()}")

all_items_in_directory = os.listdir()

myfiles = []

for item_name in all_items_in_directory:
    # Construct the full path to the item
    full_item_path = os.path.join(os.getcwd(), item_name)

    # Check if the current item is an actual file (not a directory)
    if os.path.isfile(full_item_path):
        myfiles.append(item_name) # Append the name of the file

# After the loop, print the list of files found
print("\nList of actual files in the current directory:")
for f in myfiles:
    print(f)

# Or print the entire list at once:
# print(myfiles)
```

```python
def is_power_of(base, number):
    if base <= 1 or number < 1:
        print("Invalid input.")
        return False

    print(f"Checking if {number} is a power of {base}")

    while number % base == 0:
        print(f"{number} is divisible by {base}. Dividing...")
        number = number // base
        print(f"Now number is: {number}")

    if number == 1:
        print(f"Yes! It's a power of {base} ✅")
        return True
    else:
        print(f"Nope! We ended at {number} — not a power of {base} ❌")
        return False

# Example:
is_power_of(2, 32)  # Outputs steps
```
