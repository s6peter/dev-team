```python
###this function checks if a number is a prime
def is_prime(n):
    if n < 2:
        return False
    if n ==2:
        return True
    if n % 2 ==0:
        return False
    for i in range(3, int(n**.5) + 1, 2):
        if n % i == 0:
            return False
    return True

is_prime(100)

def next_prime(n):
    candidate = n + 1
    while True:
        if is_prime(candidate):
            return candidate
        candidate +=1
        
next_prime(89)
```

```python
##classic binary search leetcode 704
def find_target(arr, target):
        left , right = 0 , len(arr)-1
        while left <= right:
            mid = (left + right) // 2
            if arr[mid]== target:
                return arr[mid]
            elif arr[mid] < target:
                left = mid +1
            else:
                right = mid -1
        return -1
        
        
arr = [1,2,3,4,5,6]
target = 1
find_target(arr, target)        
        
        
        
```
