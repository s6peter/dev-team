````python
def find_unique_string(words):
    seen = set()
    repeated = set()

    for word in words:
        if word in seen:
            repeated.add(word)  # Add duplicates to `repeated`
        else:
            seen.add(word)  # Add new words to `seen`

    # Find unique words (words that appeared only once)
    unq = list(seen - repeated)

    return unq if unq else ''  # Return the first unique word or an empty string

# Test cases


print(find_unique_string(['apple', 'banana', 'apple', 'mango', 'banana','peter', 'adam']))  # Output: 'mango'
print(find_unique_string(['hello', 'world', 'hello']))  # Output: 'world'
print(find_unique_string(['hello', 'world', 'hello', 'world']))  # Output: ''
print(find_unique_string([]))  # Output: ''
````

````python
# Splitting Log Files: Methods and Examples

## 1. Splitting by New Line (`\n`)
### Best for: Logs stored in a multi-line string.
#### Method: `split("\n")`
```python
log_data = """[2025-03-28 10:15:32] INFO - Server started
[2025-03-28 10:16:45] WARNING - High memory usage
[2025-03-28 10:17:10] ERROR - Database connection failed"""

log_lines = log_data.split("\n")
print(log_lines)  # Each log entry is now an element in the list
```

---

## 2. Splitting by Space (`split()`)
### Best for: Extracting parts of a log entry (date, time, level).
#### Method: `split(" ")`
```python
log_entry = "[2025-03-28 10:17:10] ERROR - Database connection failed"
log_parts = log_entry.split(" ")

print(log_parts)
# ['[2025-03-28', '10:17:10]', 'ERROR', '-', 'Database', 'connection', 'failed']
```

---

## 3. Splitting by a Specific Character
### Best for: Logs that use a delimiter like `-`, `|`, `:`.
#### Method: `split("-")`, `split("|")`, `split(":")`
```python
log_entry = "2025-03-28 | ERROR | Database connection failed"
log_parts = log_entry.split("|")

print(log_parts)
# ['2025-03-28 ', ' ERROR ', ' Database connection failed']
```

---

## 4. Splitting by Multiple Spaces (`re.split()`)
### Best for: Logs with variable spaces.
#### Method: `re.split(r'\s+', log_entry)`
```python
import re

log_entry = "[2025-03-28 10:17:10]   ERROR  -  Database connection failed"
log_parts = re.split(r'\s+', log_entry)

print(log_parts)
# ['[2025-03-28', '10:17:10]', 'ERROR', '-', 'Database', 'connection', 'failed']
```

---

## 5. Using Regular Expressions (`re.match()` or `re.findall()`)
### Best for: Extracting structured data like timestamps, log levels, or IPs.
#### Method: `re.findall(pattern, log_entry)`
```python
import re

log_entry = "[2025-03-28 10:17:10] ERROR - Database connection failed"
match = re.match(r"\[(.*?)\] (\w+) - (.+)", log_entry)

if match:
    timestamp, level, message = match.groups()
    print(f"Timestamp: {timestamp}, Level: {level}, Message: {message}")

# Output:
# Timestamp: 2025-03-28 10:17:10, Level: ERROR, Message: Database connection failed
```

---

## 6. Splitting by Log Format (Apache Logs Example)
### Best for: Structured logs like Apache access logs.
#### Method: `re.split(r'"\s|\s"', log_entry)`
```python
log_entry = '192.168.1.100 - - [28/Mar/2025:10:30:45 +0000] "GET /index.html HTTP/1.1" 200 5320'
log_parts = re.split(r'\s(?=(?:[^"]*"[^"]*")*[^"]*$)', log_entry)

print(log_parts)
# ['192.168.1.100', '-', '-', '[28/Mar/2025:10:30:45', '+0000]', '"GET', '/index.html', 'HTTP/1.1"', '200', '5320']
```

---

## 7. Using `csv.reader()` for CSV-style logs
### Best for: Logs stored as comma-separated values (CSV).
#### Method: `csv.reader()`
```python
import csv
from io import StringIO

log_data = """timestamp,level,message
2025-03-28 10:17:10,ERROR,Database connection failed
2025-03-28 10:19:05,INFO,Scheduled backup started"""

log_reader = csv.reader(StringIO(log_data))

for row in log_reader:
    print(row)

# Output:
# ['timestamp', 'level', 'message']
# ['2025-03-28 10:17:10', 'ERROR', 'Database connection failed']
# ['2025-03-28 10:19:05', 'INFO', 'Scheduled backup started']
```

---

## 8. Using `pandas` for Large Log Files
### Best for: Processing large log files efficiently.
#### Method: `pd.read_csv()`
```python
import pandas as pd
from io import StringIO

log_data = """timestamp,level,message
2025-03-28 10:17:10,ERROR,Database connection failed
2025-03-28 10:19:05,INFO,Scheduled backup started"""

df = pd.read_csv(StringIO(log_data))
print(df)

# Output:
#             timestamp  level                      message
# 0  2025-03-28 10:17:10  ERROR  Database connection failed
# 1  2025-03-28 10:19:05   INFO    Scheduled backup started
```

---

## Which Method Should You Use?
| **Method**                 | **Best Use Case** |
|----------------------------|------------------|
| `split("\n")`              | Multi-line logs |
| `split(" ")`               | Space-separated data |
| `split("-")`, `split("|")` | Logs with structured separators |
| `re.split(r'\s+')`         | Logs with variable spaces |
| `re.match()` / `re.findall()` | Extract specific elements (timestamps, IPs, levels) |
| `csv.reader()`             | CSV-style logs |
| `pandas.read_csv()`        | Large logs with structured format |

---

## Next Steps
Would you like to:
- **Extract specific details (IP addresses, timestamps)?**
- **Analyze log trends (error frequency, user activity)?**
- **Automate log monitoring?**

Let me know how you want to proceed! 🚀

````

````python
import string

def password_strength_counter(password):
    strength_checks = {
        'length': len(password) >= 8,
        'digit': any(char in string.digits for char in password),
        'lowercase': any(char in string.ascii_lowercase for char in password),
        'uppercase': any(char in string.ascii_uppercase for char in password),
        'special': any(char in string.punctuation for char in password),
    }

    score = sum(strength_checks.values())
    
    if score == 5:
        return f"✅ {password} is strong! (Score: {score}/5)"
    else:
        # Build suggestions
        suggestions = []
        if not strength_checks['length']:
            suggestions.append("make it at least 8 characters")
        if not strength_checks['digit']:
            suggestions.append("add a digit (0–9)")
        if not strength_checks['lowercase']:
            suggestions.append("include a lowercase letter")
        if not strength_checks['uppercase']:
            suggestions.append("include an uppercase letter")
        if not strength_checks['special']:
            suggestions.append("add a special character (!@#$...)")

        suggestion_text = "; ".join(suggestions)
        return f"⚠️ {password} is weak. (Score: {score}/5)\nSuggestions: {suggestion_text}"

# Test example
password = "57yy58"
print(password_strength_counter(password))
````

````python
def password_strength_counter(password):
    strength = {
        'length': False,
        'digit': False,
        'lowercase': False,
        'uppercase': False,
    }
    if len(password) >= 8:
        strength['length'] = True
    for char in password:
        if char.isdigit():
            strength['digit'] = True
        elif char.islower():
            strength['lowercase'] = True
        elif char.isupper():
            strength['uppercase'] = True
    return strength

# Test example
password = "57yy58"
print(password_strength_counter(password))
````

````python
Absolutely! You're looking at a simple but powerful way to manage logs on a Linux system using **logrotate**.

Let's break this into **two parts**:

---

## 🧠 1. **What is logrotate?**

**`logrotate`** is a **Linux utility** that automatically:

* 📁 **Archives**,
* 🧹 **Deletes**,
* 📦 **Compresses**, and
* 🔁 **Rotates** log files

It prevents logs from growing **too large** and filling up disk space — which is critical for production systems, especially with apps that log a lot (like web servers, Docker, syslog, etc).

---

### ✅ When to Use logrotate:

Use it when:

* Logs are growing rapidly (e.g., `/var/log/myapp.log`)
* You want to:

  * Keep only recent N days of logs
  * Compress old logs to save space
  * Avoid restarting your app just to rotate logs

---

## 🧾 2. Explanation of Your Python Code

```python
logrotate_config = """
/var/log/myapp/*.log {
    daily
    rotate 7
    compress
    missingok
    notifempty
    copytruncate
}
"""
```

This defines a **logrotate rule** for logs under `/var/log/myapp/*.log`.

### 🔍 What each directive means:

| Directive      | Meaning                                                                                           |
| -------------- | ------------------------------------------------------------------------------------------------- |
| `daily`        | Rotate logs **every day**                                                                         |
| `rotate 7`     | Keep **7 rotated logs**, delete older ones                                                        |
| `compress`     | Compress old logs (e.g., `log.1.gz`)                                                              |
| `missingok`    | Don’t error if logs are missing                                                                   |
| `notifempty`   | Skip rotation if log file is empty                                                                |
| `copytruncate` | Copy the log to a backup and **truncate** the original (used when the app **can’t be restarted**) |

---

### 🔧 Then this part:

```python
with open("/etc/logrotate.d/myapp", "w") as f:
    f.write(logrotate_config)
```

This writes your config to a file under:

```
/etc/logrotate.d/myapp
```

✅ This is the **correct directory** for custom logrotate rules on most Linux distros. The system will automatically pick it up during the daily cron job.

---

## ✅ What Happens After Setup?

Every day, `logrotate` runs (via cron or systemd timer) and:

* Finds all files matching `/var/log/myapp/*.log`
* If they’re **non-empty**, it:

  * Copies and truncates the log
  * Compresses the old one
  * Keeps only the last 7

---

### 📦 Example Result:

```
/var/log/myapp/myapp.log
/var/log/myapp/myapp.log.1.gz
/var/log/myapp/myapp.log.2.gz
...
/var/log/myapp/myapp.log.7.gz
```

---

## ✅ TL;DR Summary

| What it is     | Why use it                       |
| -------------- | -------------------------------- |
| `logrotate`    | Auto-manages logs                |
| `rotate 7`     | Keep 7 old copies                |
| `compress`     | Save disk space                  |
| `copytruncate` | Keeps app running while rotating |
| Python code    | Creates the config file          |

---

Let me know if you want:

* A version that rotates hourly or weekly
* To manually trigger it with `logrotate -f`
* A way to use it in Docker or Kubernetes environments
````

````python



def make_shortest_palindrome(s):
    """
    Returns the shortest palindrome that can be formed by adding characters in front of 's'.
    Case-insensitive logic for matching; output preserves original case.
    """
    original = s
    s_lower = s.lower()  # Normalize for comparison

    if s_lower == s_lower[::-1]:
        return original  # Already a palindrome

    for i in range(len(s_lower)):
        suffix = s_lower[i:]
        if suffix == suffix[::-1]:  # Found longest palindromic suffix
            prefix_to_add = original[:i][::-1]  # Use original string to preserve case
            return prefix_to_add + original

    return original[::-1] + original  # Fallback

# Example usage
input_str = "aacecaaa"
output_str = make_shortest_palindrome(input_str)
print("Input: ", input_str)
print("Output:", output_str)
````
