````python
def find_unique_string(words):
    seen = set()
    repeated = set()

    for word in words:
        if word in seen:
            repeated.add(word)  # Add duplicates to `repeated`
        else:
            seen.add(word)  # Add new words to `seen`

    # Find unique words (words that appeared only once)
    unq = list(seen - repeated)

    return unq[-1] if unq else ''  # Return the first unique word or an empty string

# Test cases
print(find_unique_string(['apple', 'banana', 'apple', 'mango', 'banana']))  # Output: 'mango'
print(find_unique_string(['hello', 'world', 'hello']))  # Output: 'world'
print(find_unique_string(['hello', 'world', 'hello', 'world']))  # Output: ''
print(find_unique_string([]))  # Output: ''


from collections import defaultdict

def find_anagram_words(list_1, list_2):
    my_dict1 = defaultdict(list)
    my_dict2 = defaultdict(list)

    # Create dictionary for list_1
    for w1 in list_1:
        sorted_word = "".join(sorted(w1.lower()))  # Key is the sorted word
        my_dict1[sorted_word].append(w1)  # Store original words

    # Create dictionary for list_2
    for w2 in list_2:
        sorted_word = "".join(sorted(w2.lower()))  # Key is the sorted word
        my_dict2[sorted_word].append(w2)

    # Find common anagram groups
    result = []
    for key in my_dict1:
        if key in my_dict2:
            result.extend(my_dict1[key])  # Append matching words

    return result  # Return final list of matching words

# Test cases
print(find_anagram_words(['cinema', 'iceman'], ['iceman', 'cinema']))  # ['cinema', 'iceman']
print(find_anagram_words(['test', 'stet'], ['tent', 'nett']))  # []
print(find_anagram_words(['hello', 'world'], ['dolly', 'sir']))  # []
````

````python
from collections import defaultdict
from datetime import datetime

# Create a defaultdict where each log type stores a list of log messages
log_system = defaultdict(list)

def log_event(log_type, message):
    """Adds a log entry under the specified log type."""
    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    log_system[log_type].append(f"[{timestamp}] {message}")

# Logging different types of events
log_event("INFO", "System startup complete.")
log_event("WARNING", "CPU usage is high.")
log_event("ERROR", "Database connection failed.")
log_event("INFO", "Scheduled backup completed.")
log_event("ERROR", "Failed to write to disk.")

# Print all logs grouped by category
for log_type, messages in log_system.items():
    print(f"\n{log_type} LOGS:")
    for msg in messages:
        print(f"  - {msg}")
        
from collections import defaultdict

# Dictionary to categorize system metrics
metrics = defaultdict(list)

def record_metric(category, value):
    """Records a new system metric under the given category."""
    metrics[category].append(value)

# Simulating system metric recordings
record_metric("CPU Usage", 75)
record_metric("CPU Usage", 80)
record_metric("Memory Usage", 60)
record_metric("Memory Usage", 65)
record_metric("Disk Usage", 50)

# Print system metrics
for category, values in metrics.items():
    print(f"\n{category}:")
    print(f"  - {values} (Avg: {sum(values)/len(values):.2f}%)")

````

````python
# Splitting Log Files: Methods and Examples

## 1. Splitting by New Line (`\n`)
### Best for: Logs stored in a multi-line string.
#### Method: `split("\n")`
```python
log_data = """[2025-03-28 10:15:32] INFO - Server started
[2025-03-28 10:16:45] WARNING - High memory usage
[2025-03-28 10:17:10] ERROR - Database connection failed"""

log_lines = log_data.split("\n")
print(log_lines)  # Each log entry is now an element in the list
```

---

## 2. Splitting by Space (`split()`)
### Best for: Extracting parts of a log entry (date, time, level).
#### Method: `split(" ")`
```python
log_entry = "[2025-03-28 10:17:10] ERROR - Database connection failed"
log_parts = log_entry.split(" ")

print(log_parts)
# ['[2025-03-28', '10:17:10]', 'ERROR', '-', 'Database', 'connection', 'failed']
```

---

## 3. Splitting by a Specific Character
### Best for: Logs that use a delimiter like `-`, `|`, `:`.
#### Method: `split("-")`, `split("|")`, `split(":")`
```python
log_entry = "2025-03-28 | ERROR | Database connection failed"
log_parts = log_entry.split("|")

print(log_parts)
# ['2025-03-28 ', ' ERROR ', ' Database connection failed']
```

---

## 4. Splitting by Multiple Spaces (`re.split()`)
### Best for: Logs with variable spaces.
#### Method: `re.split(r'\s+', log_entry)`
```python
import re

log_entry = "[2025-03-28 10:17:10]   ERROR  -  Database connection failed"
log_parts = re.split(r'\s+', log_entry)

print(log_parts)
# ['[2025-03-28', '10:17:10]', 'ERROR', '-', 'Database', 'connection', 'failed']
```

---

## 5. Using Regular Expressions (`re.match()` or `re.findall()`)
### Best for: Extracting structured data like timestamps, log levels, or IPs.
#### Method: `re.findall(pattern, log_entry)`
```python
import re

log_entry = "[2025-03-28 10:17:10] ERROR - Database connection failed"
match = re.match(r"\[(.*?)\] (\w+) - (.+)", log_entry)

if match:
    timestamp, level, message = match.groups()
    print(f"Timestamp: {timestamp}, Level: {level}, Message: {message}")

# Output:
# Timestamp: 2025-03-28 10:17:10, Level: ERROR, Message: Database connection failed
```

---

## 6. Splitting by Log Format (Apache Logs Example)
### Best for: Structured logs like Apache access logs.
#### Method: `re.split(r'"\s|\s"', log_entry)`
```python
log_entry = '192.168.1.100 - - [28/Mar/2025:10:30:45 +0000] "GET /index.html HTTP/1.1" 200 5320'
log_parts = re.split(r'\s(?=(?:[^"]*"[^"]*")*[^"]*$)', log_entry)

print(log_parts)
# ['192.168.1.100', '-', '-', '[28/Mar/2025:10:30:45', '+0000]', '"GET', '/index.html', 'HTTP/1.1"', '200', '5320']
```

---

## 7. Using `csv.reader()` for CSV-style logs
### Best for: Logs stored as comma-separated values (CSV).
#### Method: `csv.reader()`
```python
import csv
from io import StringIO

log_data = """timestamp,level,message
2025-03-28 10:17:10,ERROR,Database connection failed
2025-03-28 10:19:05,INFO,Scheduled backup started"""

log_reader = csv.reader(StringIO(log_data))

for row in log_reader:
    print(row)

# Output:
# ['timestamp', 'level', 'message']
# ['2025-03-28 10:17:10', 'ERROR', 'Database connection failed']
# ['2025-03-28 10:19:05', 'INFO', 'Scheduled backup started']
```

---

## 8. Using `pandas` for Large Log Files
### Best for: Processing large log files efficiently.
#### Method: `pd.read_csv()`
```python
import pandas as pd
from io import StringIO

log_data = """timestamp,level,message
2025-03-28 10:17:10,ERROR,Database connection failed
2025-03-28 10:19:05,INFO,Scheduled backup started"""

df = pd.read_csv(StringIO(log_data))
print(df)

# Output:
#             timestamp  level                      message
# 0  2025-03-28 10:17:10  ERROR  Database connection failed
# 1  2025-03-28 10:19:05   INFO    Scheduled backup started
```

---

## Which Method Should You Use?
| **Method**                 | **Best Use Case** |
|----------------------------|------------------|
| `split("\n")`              | Multi-line logs |
| `split(" ")`               | Space-separated data |
| `split("-")`, `split("|")` | Logs with structured separators |
| `re.split(r'\s+')`         | Logs with variable spaces |
| `re.match()` / `re.findall()` | Extract specific elements (timestamps, IPs, levels) |
| `csv.reader()`             | CSV-style logs |
| `pandas.read_csv()`        | Large logs with structured format |

---

## Next Steps
Would you like to:
- **Extract specific details (IP addresses, timestamps)?**
- **Analyze log trends (error frequency, user activity)?**
- **Automate log monitoring?**

Let me know how you want to proceed! 🚀

````

````python

import subprocess

username = "selina"
default_password = "SecurePass123"  # Change this to your preferred default password

def user_exists(username):
    """Check if a user exists in /etc/passwd"""
    try:
        subprocess.run(["id", username], check=True, capture_output=True, text=True)
        return True
    except subprocess.CalledProcessError:
        return False

def create_user(username, password):
    """Create a user, set a password, and add to a group"""
    print(f"Creating user: {username}")
    
    # Create the user
    subprocess.run(["sudo", "useradd", "-m", "-s", "/bin/bash", username], check=True)
    
    # Set password for user
    subprocess.run(["sudo", "bash", "-c", f"echo '{username}:{password}' | chpasswd"], check=True)
    
    # Add user to 'hr' group
    #subprocess.run(["sudo", "usermod", "-aG", "hr", username], check=True)
    
    print(f"User {username} created and added to 'hr' group.")

def display_user_info(username):
    """Display user details"""
    print(f"\nUser details for {username}:")
    subprocess.run(["id", username])
    subprocess.run(["grep", f"^{username}:", "/etc/passwd"])
    subprocess.run(["sudo", "grep", f"^{username}:", "/etc/shadow"])

if __name__ == "__main__":
    if user_exists(username):
        print(f"The user '{username}' already exists.")
    else:
        create_user(username, default_password)
        print(f"User '{username}' created successfully with default password.")

    display_user_info(username)
````
