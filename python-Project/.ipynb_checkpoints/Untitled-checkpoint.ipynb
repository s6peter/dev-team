```python
import subprocess as sp
sp.run(['ls','-l'])
```

```python
result = sp.run(['ls', '-l'], capture_output=True, text=True)
print(result.stdout)
```

```python
with open('output.txt', 'w') as f:
    sp.run(['ls', '-l'], stdout=f)
sp.run(['ls'])
sp.run(['cat', 'output.txt'])
```

```python
process = sp.Popen(['ping', '-c', '4', 'google.com'], stdout=sp.PIPE, stderr=sp.PIPE, text=True)

# Read output line by line (useful for long-running processes)
for line in process.stdout:
    print(line.strip())

process.wait()  # Ensure the process completes
```

```python
import subprocess as sp

# List files in the current directory
sp.run(['ls'])

# Change directory to './v-project' and run 'ls'
sp.run(['ls'], cwd='./v-project')
```

```python
sp.run(['sudo', 'apt', 'update'])
sp.run('cat', input='Hello, world!', text=True, shell=True)

```

```python
p1 = sp.Popen(['cat', 'file.txt'], stdout=sp.PIPE, text=True)
p2 = sp.Popen(['grep', 'hello'], stdin=p1.stdout, stdout=sp.PIPE, text=True)

output = p2.communicate()[0]
print(output)
```
